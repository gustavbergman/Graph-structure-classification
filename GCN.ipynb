### Convert to PyTorch geometric
### create node features

```python
def create_node_features(G):

    degrees = dict(G.degree())
    clustering = nx.clustering(G)
    betweenness = nx.betweenness_centrality(G)

    x = []

    for node in G.nodes():

        x.append([
            degrees[node],
            clustering[node],
            betweenness[node]
        ])

    return torch.tensor(x, dtype=torch.float)



def graph_to_data(G, label):

    data = from_networkx(G)
    data.x = create_node_features(G)
    data.y = torch.tensor([label], dtype=torch.long)

    return data


# G = generate_watts_strogatz(30)
# data = graph_to_data(G_ws, 0)

# print(data)
# print(data.x.shape)
# print(data.edge_index.shape)
```

### build dataset

```python
def generate_dataset(num_graphs_per_class=500):

    dataset = []

    for _ in range(num_graphs_per_class):

        # Watts-Strogatz
        G = generate_watts_strogatz(
            np.random.randint(20, 100)
        )

        dataset.append(graph_to_data(G, 0))


        # Barabasi-Albert
        G = generate_barabasi_albert(
            np.random.randint(20, 100)
        )

        dataset.append(graph_to_data(G, 1))


        # Random Regular
        G = generate_random_regular(
            np.random.randint(20, 100)
        )

        dataset.append(graph_to_data(G, 2))

    return dataset
```

### generate dataset

```python
dataset = generate_dataset(500)

print(len(dataset))

print(dataset[0])
print(dataset[100])
```

### training and test splits

```python
train, test = train_test_split(
    dataset,
    test_size=0.2,
    random_state=42,
    shuffle=True
)
```

```python
train_loader = DataLoader(
    train,
    batch_size=32,
    shuffle=True
)

test_loader = DataLoader(
    test,
    batch_size=32
)
```

```python
batch = next(iter(train_loader))
print(batch)
```

# GCN

```python
import torch
import torch.nn.functional as F
from torch.nn import Linear

from torch_geometric.nn import GCNConv
from torch_geometric.nn import global_mean_pool
from torch_geometric.nn import global_add_pool
from torch_geometric.nn import BatchNorm
```

```python
class GCN(torch.nn.Module):

    def __init__(self):

        super().__init__()

        # Graph convolution layers
        self.conv1 = GCNConv(3, 64)
        self.conv2 = GCNConv(64, 64)
        self.conv3 = GCNConv(64, 64)

        self.bn1 = BatchNorm(64)
        self.bn2 = BatchNorm(64)
        self.bn3 = BatchNorm(64)

        # Final classifier
        self.lin = Linear(64, 3)



    def forward(self, x, edge_index, batch):

        # First graph convolution
        x = self.conv1(x, edge_index)
        x = self.bn1(x)
        x = F.relu(x)
        x = F.dropout(x, p=0.3, training=self.training)

        # Second graph convolution
        x = self.conv2(x, edge_index)
        x = self.bn2(x)
        x = F.relu(x)
        x = F.dropout(x, p=0.3, training=self.training)

        x = self.conv3(x, edge_index)
        x = self.bn3(x)
        x = F.relu(x)

        # Convert node embeddings --> graph embedding
        x = global_add_pool(x, batch)

        # Final classification layer
        x = self.lin(x)

        return x
```

### initialize model

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = GCN().to(device)
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0001
)

loss_function = torch.nn.CrossEntropyLoss()
```

### training function

```python
def train():

    model.train()

    total_loss = 0

    for data in train_loader:

        data = data.to(device)

        optimizer.zero_grad()

        out = model(
            data.x,
            data.edge_index,
            data.batch
        )

        loss = loss_function(out, data.y)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    return total_loss
```

### evaluation function

```python
def test(loader):

    model.eval()

    correct = 0

    for data in loader:

        data = data.to(device)

        out = model(
            data.x,
            data.edge_index,
            data.batch
        )

        pred = out.argmax(dim=1)

        correct += int((pred == data.y).sum())

    return correct / len(loader.dataset)
```

### run model

```python
train_accs = []
test_accs = []
losses = []

best_test_acc = 0
best_epoch = 0

for epoch in range(1, 51):

    loss = train()

    train_acc = test(train_loader)

    test_acc = test(test_loader)

    # appends
    losses.append(loss)
    train_accs.append(train_acc)
    test_accs.append(test_acc)

    # Save best model
    if test_acc > best_test_acc:

        best_test_acc = test_acc
        best_epoch = epoch

        # torch.save(
        #     model.state_dict(),
        #     "best_gcn.pt"
        # )

    print(
        f"Epoch {epoch:3d} | "
        f"Loss {loss:.3f} | "
        f"Train {train_acc:.3f} | "
        f"Test {test_acc:.3f}"
    )

print("\nBest Test Accuracy:", best_test_acc)
print("Best Epoch:", best_epoch)
```

### confusion matrix

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay


y_true = []
y_pred = []

model.eval()

for data in test_loader:

    data = data.to(device)

    out = model(
        data.x,
        data.edge_index,
        data.batch
    )

    pred = out.argmax(dim=1)

    y_true.extend(data.y.cpu().numpy())
    y_pred.extend(pred.cpu().numpy())


cm = confusion_matrix(y_true, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Watts-Strogatz",
        "Barabasi-Albert",
        "Random Regular"
    ]
)

disp.plot(cmap="Blues")

plt.show()
```
